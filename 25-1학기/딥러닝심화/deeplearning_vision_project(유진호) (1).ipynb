# YOLOv8 기반 객체 탐지 실습 보고서

## 1. 실험 목적
- 수업시간에 배운 YOLO 모델을 활용하여 실제 데이터를 학습해보고, 결과를 시각화 및 분석

## 2. 사용 모델 및 데이터
- 모델: YOLOv8n (ultralytics)
- 데이터: 직접 수집/주석한 이미지 144장 (train: 124장, val: 20장)
- 클래스: 1개 (예: object)

## 3. 학습 환경
- Google Colab
- Python, Ultralytics YOLOv8

## 4. 학습 설정
- Epoch: 5
- Batch Size: 8
- Image Size: 640x640

## 5. 학습 결과(0번 객체만)
- Precision: 0.867
- Recall: 0.51
- mAP@50: 0.689
- mAP@50-95: 0.392
